## General Scenario for the resit

In the resit, we will work with a simulated dataset from an online store that created a set of campaigns. This dataset will be used to predict how much someone has spent in the webshop (i.e.,```order_euros```). 

Scenario:
Our webshop has launched new campaigns to increase revenue (```order_euros```). 

We are interested in two campaigns (indicated in column ```type_campaign```):
* The CPC campaign running on Facebook and Google
* The SEA campaign running on Google

We want to know if each campaign led to an increase in revenue compared to the other campaigns (i.e., any traffic source that is not set as CPC or SEA).

The dependent variable (or target) is ```order_euros```. Your independent variables (or features) will be relevant characteristics available on (or created from) the dataset.

For the resit, you will use a dataset used in class in week 6. It contains all variables necessary (and thus does not require merging). You can find it in the Github folder; it is called ```da6_fulldata.csv.gz``` (for csv file) or ```da6_fulldata.pkl.gz``` (for pickle file). It contains information about:
* sessions. Columns ```session_id```, ```session_timestamp```, ```user_agent```, ```referral```, ```type_campaign``` tell us  the details of the browser that visited the site (user_agent), when the visit started, what the referral was, and what campaign the referral came from.
* orders. Columns ```order_euros``` and ```purchase``` include information if a user made a purchase and how much they spent.
* users. Columns ```user_id```, ```reg_name```, ```registration_date```, ```initial_referrer```, ```preferential_client``` and ```age``` for registered users tell us their name, when they created their profile, their age, the referral that they followed when creating their profile, and whether they are part of an exclusive club of preferential clients
