### Now we know that we can roll a dice in python (yay!), but how should we calculate scores? Check out the different options below and give them names.

```python
#scoring option 1: what is the sum of all numbers on your dice?
def score_sum(rolls):
    return sum(rolls)

# BUG: cannot add an integer to a list; need sum(rolls); the + 1 also does not make sense - why would you want to add a random 1 to your score?
# Note: This cell does not give you any output as it only defines the function!
```

```python
#scoring option 2: are all dice the same?
def score_all_same(rolls):
    if len(set(rolls)) == 1:
        return True
    else:
        return False

# BUG: if len(set(rolls)) > 1; logic reversed; >1 means NOT all dice are the same; corrected to == 1
# set = number of unique values in 'rolls'. If 1, then only one value is in rolls, so all values are the same (one unique value)
# Note: This cell does not give you any output as it only defines the function!
```

```python
#scoring option 3: what is the highest number on your dice?
def score_highest(rolls):
    return max(rolls)

# Note: This cell does not give you any output as it only defines the function!
```

### Alright! And now let's see what the score is for your roll :)

```python
#choosing a set scoring rule
choice = 1 #you can change the number here depending on what scoring option you want to use.

# defining what is done for each choice (e.g., 1 = summarise all numbers you rolled)
if choice == 1:
    score = score_sum(rolls)
elif choice == 2:
    score = score_all_same(rolls)
elif choice == 3:
    score = score_highest(rolls)

# BUG: use == for comparison, = is assignment (added for if choice = 1)
# Note: This cell does not give you any output as it only defines conditions!
```

```python
#showing the score
print("Your score is:", score) # printing score on the screen, according to the scoring option you have chosen in the cell before.
```
